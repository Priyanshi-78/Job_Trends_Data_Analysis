```python
df["Min_Salary"] = pd.to_numeric(df["Min_Salary"], errors="coerce")
df["Max_Salary"] = pd.to_numeric(df["Max_Salary"], errors="coerce")
```

1. Find the top 10 companies with the highest number of job postings.

```python
top_company = df["Company"].value_counts().head(10)
print(top_company)
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=top_company.values, y=top_company.index, palette="viridis")

plt.title("Top 10 Companies")
plt.xlabel("Job Postings")
plt.ylabel("Company")
plt.show()
```

2. Which job title has the highest number of openings? Display the top 10 job titles.

```python
top_job = df["Job_Title"].value_counts().head(10)
print(top_job)
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=top_job.values, y=top_job.index, palette="magma")

plt.title("Top 10 Job Titles")
plt.xlabel("Number of Openings")
plt.ylabel("Job Title")
plt.show()
```

3. Find the top 10 locations with the maximum number of job postings.

```python
top_location = df["Location"].value_counts().head(10)
print(top_location)
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=top_location.values, y=top_location.index, palette="Set2")

plt.title("Top 10 Locations")
plt.xlabel("Job Postings")
plt.ylabel("Location")
plt.show()
```

4. Calculate the average, minimum, and maximum salary for each job title.

```python
salary = df.groupby("Job_Title")[["Min_Salary","Max_Salary"]].agg(["mean","min","max"])

print(salary)
```

```python
salary = df.groupby("Job_Title")[["Min_Salary","Max_Salary"]].mean().reset_index()

top10 = salary.sort_values("Max_Salary", ascending=False).head(10)

plt.figure(figsize=(10,5))

top10.plot(
    x="Job_Title",
    y=["Min_Salary","Max_Salary"],
    kind="bar",
    colormap ="Dark2"
)

plt.title("Average Min Salary and Average Max Salary")
plt.xlabel("Job Title")
plt.ylabel("Salary")
plt.xticks(rotation=90)
plt.show()
```

5. Which company offers the highest average maximum salary?

```python
company_salary = df.groupby("Company")["Max_Salary"].mean().sort_values(ascending=False)

print(company_salary.head(10))
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=company_salary.head(10).values,
            y=company_salary.head(10).index,
            palette="plasma")

plt.title("Top 10 Companies by Average Max Salary")
plt.xlabel("Average Max Salary")
plt.ylabel("Company")
plt.show()
```

6. Find all job postings where the salary range (Max Salary - Min Salary) is greater than ₹10,00,000.

```python
df["Salary_Range"] = df["Max_Salary"] - df["Min_Salary"]

job = df[df["Salary_Range"] > 1000000] \
    .groupby("Job_Title")["Salary_Range"] \
    .max() \
    .sort_values(ascending=False) \
    .head(10)

print(job)
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=job.values,
    y=job.index,
    palette="crest"
)

plt.title("Top 10 Job Postings Where Salary Range > ₹10,00,000")
plt.xlabel("Salary Range")
plt.ylabel("Job Title")
plt.show()
```

7. Count the number of job postings created on each job creation date.

```python
count=df.groupby('Job_Title')['Job_Created_Date'].count().sort_values(ascending=False).head(10)
print(count)
```

```python
plt.figure(figsize=(6,6))
sns.barplot(data=count,palette='rocket')
plt.title('Number of Job Postings',fontweight='bold',color='Purple')
plt.xticks(rotation=90)
plt.show()
```

8. Find the top 10 highest-paying jobs based on the maximum salary.

```python
top_salary = df.nlargest(10,"Max_Salary")[["Job_Title","Company","Location","Max_Salary"]]
print(top_salary)
```

```python
plt.figure(figsize=(8,5))
sns.barplot(data=top_salary,
            x="Max_Salary",
            y="Job_Title",
            palette="rocket")

plt.title("Top 10 Highest Paying Jobs")
plt.xlabel("Maximum Salary")
plt.ylabel("Job Title")
plt.show()
```

9. Calculate the average salary range (Max Salary - Min Salary) for each company.

```python
df["Salary_Range"] = df["Max_Salary"] - df["Min_Salary"]

avg_range = df.groupby("Company")["Salary_Range"].mean().sort_values(ascending=False)

print(avg_range)
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=avg_range.head(10).values,
            y=avg_range.head(10).index,
            palette="cubehelix")

plt.title("Top 10 Companies by Average Salary Range")
plt.xlabel("Average Salary Range")
plt.ylabel("Company")
plt.show()
```

10. Which location has the highest average salary? Display the top 10 locations by average salary.

```python
df["Average_Salary"] = (df["Min_Salary"] + df["Max_Salary"]) / 2

location_salary = df.groupby("Location")["Average_Salary"].mean().sort_values(ascending=False)

print(location_salary.head(10))
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x=location_salary.head(10).values,
            y=location_salary.head(10).index,
            palette="coolwarm")

plt.title("Top 10 Locations by Average Salary")
plt.xlabel("Average Salary")
plt.ylabel("Location")
plt.show()
```
